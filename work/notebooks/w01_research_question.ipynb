### Selected Lane: Lane 2 — Content Refresh Opportunity Scoring

**Why this lane:** Websites with thousands of pages face a huge challenge: search traffic drops quietly, and by the time a page loses all traffic, it takes months to recover. Managing updates manually or randomly refreshing old content wastes editorial time and budget. I chose Lane 2 to build a Machine Learning priority scoring system that scans past performance signals and identifies decay-risk pages early, creating an actionable review queue for the content team.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

* **Unit of Analysis (Grain):** A single indexed web page URL observed over search performance time windows.
* **The Search Question:** "Which web pages are at high risk of organic search traffic decay and should be prioritized for content updates?"
* **The Decision It Improves:** Allocating writer time and SEO budget to pages that actually need updates instead of guessing randomly.
* **The Action Taken:** The content/SEO manager takes the top-ranked URLs from the model's priority queue and assigns them to writers to refresh meta tags, update facts, and fix internal links.
* **Cost of a Wrong Call:**
  * **False Alarm (False Positive):** Wasting editorial budget and writer time updating a healthy page that didn't need any fix.
  * **Missed Decay (False Negative):** A declining page goes unnoticed, causing cumulative loss in search impressions, clicks, and revenue.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

dataset_url = "https://raw.githubusercontent.com/MuhammadOkasha004/flyrank-ml-internship-work/main/data/raw/content_refresh_anonymized.csv"

# Load dataset into Pandas DataFrame
df = pd.read_csv(dataset_url)

# Metric 1: Total Units of Analysis (Total Web Pages)
total_pages = len(df)

# Metric 2: Pages facing organic traffic decay (Target Label = down)
declining_mask = (df['trend_direction'] == 'down')
declining_count = int(np.sum(declining_mask))
declining_pct = (declining_count / total_pages) * 100

# Metric 3: Recent Traffic Drop Signal (impressions_last_30d < impressions_prev_30d)
recent_drop_mask = (df['impressions_last_30d'] < df['impressions_prev_30d'])
recent_drop_count = int(np.sum(recent_drop_mask))

# Overlap precision: Recent drop pages that are officially labeled as 'down'
overlap_count = int(np.sum(recent_drop_mask & declining_mask))
signal_precision = (overlap_count / recent_drop_count) * 100 if recent_drop_count > 0 else 0.0

# Metric 4: Average 90-Day Impressions across all pages
avg_impressions = float(np.mean(df['impressions_90d'])) if 'impressions_90d' in df.columns else 0.0

# Print clean output metrics
print("==================================================================")
print("             STARTER DATASET PROOF METRICS (LANE 2)               ")
print("==================================================================")
print("1. Total Analyzed Web Pages (URLs)       :", total_pages)
print("2. Decaying Pages (Target = down)        :", declining_count, f"({declining_pct:.2f}%)")
print("3. Pages Showing Recent 30d Drop         :", recent_drop_count)
print("4. Early Drop Signal Accuracy            :", f"{round(signal_precision, 2)}%")
print("5. Average 90-Day Impressions per URL    :", round(avg_impressions, 2))
print("==================================================================")
```

### 4. Scope & Bounds: What I Can and Can't Claim

#### What this work CAN claim:
- **Directional Risk Signals:** It can identify early warning patterns (like rank position drop, lower scroll rate, and old content) that show a web page is losing traffic.
- **Smart Prioritization:** It gives every web page a Risk Score (from 0 to 1) so the content team knows exactly which pages to update first.
- **Data-Backed Decisions:** It replaces guessing with real statistical data from past performance.

#### What this work CANNOT claim:
- **No Causal Proof:** The model shows correlation, not direct cause. (e.g., It flags a dropping page, but cannot prove if it's due to Google updates or competitors).
- **Not Predicting Google's Algorithm:** It does not predict or reverse-engineer Google's secret ranking formulas. It only monitors past traffic metrics.
- **No 100% Guarantee:** The model outputs probabilities (risks), not guaranteed predictions. External factors like unexpected holidays can always create exceptions.
