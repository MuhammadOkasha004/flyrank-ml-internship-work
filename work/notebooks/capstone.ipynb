3. Methodology
Rolling-Window Formulation: The system uses a 90-day rolling-window framework to forecast the future trajectory of individual web page performance classified into three discrete outcomes: DOWN, FLAT, or UP.

Feature Engineering: Features are constructed strictly from data available at or before the cutoff date. Key inputs include core GSC metrics (impressions, clicks, CTR, average position), engagement signals (sessions), page metadata (word count, content age, freshness), and derived momentum/velocity metrics.

Leakage Control Audit: All future-period variables (such as future impression counts and post-cutoff percentage changes) are strictly identified and stripped from model feature matrices.

Human Baseline Comparison: A predefined FlyRank Human Baseline—constructed from recent-versus-prior impression shifts—is evaluated on the exact same locked test period to benchmark machine learning gains over traditional rules-based SEO methods.

Time-Aware Model Validation:

Train/Validation Split: Historical 2025 data is used for model development and hyperparameter selection.

Locked Test Evaluation: January 2026 is preserved as a completely untouched holdout test period.

Cross-Temporal Backtesting: Multi-period cross-validation is performed across earlier 2025 windows to verify temporal stability across changing search cycles.

Evaluation Framework: Beyond standard accuracy, performance is evaluated using metrics tailored to high-stakes SEO refresh decisions:

Macro F1 & Balanced Accuracy: To ensure balanced evaluation across imbalanced class distributions.

DOWN Recall & Precision: To verify the system accurately catches declining pages without over-flagging stable ones.

Severe Directional Error Rate: Minimizing catastrophic misclassifications (e.g., predicting UP for a page actually going DOWN).

High-Confidence Reliability: Measuring accuracy within top-tier probability outputs for queue prioritization.

## 5. Limitations

*What this work cannot claim.*


### 5. Limitations

* The evaluation covers a specific historical dataset and does not establish performance across all future websites or SEO environments.
* January 2026 is one future test period; longer-term production monitoring is still required.
* The model predicts directional performance, not guaranteed traffic or revenue outcomes.
* Concept drift may occur as search algorithms, competitors, content, and user behavior change.
* Low-volume pages can produce unstable percentage changes and remain difficult to classify reliably.
* FLAT remains substantially harder to distinguish than DOWN in several evaluations.
* High-confidence predictions cover only a small portion of predictions, so confidence scores should support—not replace—human review.
* The system should therefore be treated as a decision-support / early-warning system, not an autonomous SEO decision maker.
